
*	Downloads micromamba (conda-lite).
*   Creates a separate env py37 with Python 3.7 (independent of Colab’s 3.12).
*   Verifies the env.



```python
%%bash
# Install micromamba (lightweight conda)
wget -qO- https://micro.mamba.pm/api/micromamba/linux-64/latest | tar -xvj bin/micromamba
# Execute the remaining commands using the direct path to micromamba
# ./bin/micromamba shell init -s bash
./bin/micromamba create -y -r /root/micromamba -n py37 -c conda-forge python=3.7 pip
./bin/micromamba run -r /root/micromamba -n py37 python -V
```

* Installs dependencies into py37 (not the kernel).
*	Confirms zrp imports.
* Runs zrp's CLI "download" step to fetch its data/assets.


```python
%%bash
set -euo pipefail

# install pinned ZRP package version
./bin/micromamba run -r /root/micromamba -n py37 pip install zrp==0.4.1

# quick import test (won't fail the cell if zrp lacks CLI)
./bin/micromamba run -r /root/micromamba -n py37 python -c "import zrp; print('zrp ok')"

./bin/micromamba run -r /root/micromamba -n py37 python -m zrp download
```
